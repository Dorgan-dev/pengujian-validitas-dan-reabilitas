PENGUJIAN VALIDITAS DAN RELIABILITAS INSTRUMEN

Mata Kuliah Statistika Probabilitas

```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

```python
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

```python
# Hanya ambil kolom yang berformat "Dimensi [pernyataan]"
df = pd.read_excel(nama_file)
butir_cols = [c for c in df.columns if '[' in c]
butir = df[butir_cols]

print("Jumlah responden :", len(df))
print("Jumlah butir     :", len(butir_cols))
print("Data kosong      :", int(butir.isna().sum().sum()))
print("\nKolom butir yang terbaca:")
for i, c in enumerate(butir_cols, 1):
    print(f"  {i}. {c}")
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)"        : 2,
    "Netral (N)"               : 3,
    "Setuju (S)"               : 4,
    "Sangat Setuju (SS)"       : 5,
}
num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

gagal = int(num.isna().sum().sum())
print(f"Sel yang gagal dikonversi: {gagal}")
if gagal > 0:
    print("Ada jawaban yang tidak terbaca! Periksa penulisan label pada data.")
else:
    print("Semua sel berhasil dikonversi.")
num.head()
```

```python
dimensi = {}
for kolom in butir.columns:
    nama = re.match(r"^(.*?) \[", kolom).group(1).strip()
    dimensi.setdefault(nama, []).append(kolom)

print(f"Jumlah dimensi: {len(dimensi)}")
for nama, daftar in dimensi.items():
    print(f"  • {nama}: {len(daftar)} butir")
```

```python
n = len(num)
df_stat = n - 2
t_crit  = stats.t.ppf(0.975, df_stat)          # dua sisi, α=0.05
r_tabel = t_crit / np.sqrt(df_stat + t_crit**2)

print(f"n = {n}")
print(f"df = {df_stat}")
print(f"t kritis (α=0,05; dua sisi) = {t_crit:.4f}")
print(f"r tabel = {r_tabel:.3f}")
```

```python
hasil_validitas = []

for nama, daftar in dimensi.items():
    x     = num[daftar]
    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):
        r,         p  = stats.pearsonr(x[kolom], total)
        r_koreksi, _  = stats.pearsonr(x[kolom], total - x[kolom])  # corrected item-total

        hasil_validitas.append({
            "Dimensi"    : nama,
            "Butir"      : f"B{i}",
            "r hitung"   : round(r, 3),
            "r tabel"    : round(r_tabel, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value"    : round(p, 4),
            "Keterangan" : "Valid" if r > r_tabel else "Tidak Valid",
        })

tabel_validitas = pd.DataFrame(hasil_validitas)

print(f"\n{'='*60}")
print(f"TABEL HASIL VALIDITAS  (r tabel = {r_tabel:.3f})")
print(f"{'='*60}")
print(tabel_validitas.to_string(index=False))

valid_count   = (tabel_validitas["Keterangan"] == "Valid").sum()
invalid_count = (tabel_validitas["Keterangan"] == "Tidak Valid").sum()
print(f"\nButir Valid   : {valid_count}")
print(f"Butir Tidak Valid : {invalid_count}")
```

```python
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori_alpha(a):
    if   a >= 0.90: return "Sangat Baik"
    elif a >= 0.80: return "Baik"
    elif a >= 0.70: return "Reliabel"
    else:           return "Tidak Reliabel"

hasil_reliabilitas = []
alpha_dihapus      = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)

    hasil_reliabilitas.append({
        "Dimensi"       : nama,
        "Jumlah Butir"  : len(daftar),
        "Cronbach Alpha": round(a, 3),
        "Kategori"      : kategori_alpha(a),
    })

    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            a_del = cronbach_alpha(x.drop(columns=kolom))
            alpha_dihapus.append({
                "Dimensi"           : nama,
                "Butir"             : f"B{i}",
                "Alpha Dimensi"     : round(a, 3),
                "Alpha jika Dihapus": round(a_del, 3),
                "Interpretasi"      : "Melemahkan" if a_del > a else "Mempertahankan/Menguatkan",
            })

tabel_reliabilitas  = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)

print(f"\n{'='*60}")
print("TABEL HASIL RELIABILITAS")
print(f"{'='*60}")
print(tabel_reliabilitas.to_string(index=False))

print(f"\n{'='*60}")
print("ALPHA JIKA BUTIR DIHAPUS")
print(f"{'='*60}")
print(tabel_alpha_dihapus.to_string(index=False))
```

```python
gabung = tabel_validitas.merge(
    tabel_alpha_dihapus[["Dimensi", "Butir", "Alpha jika Dihapus", "Interpretasi"]],
    on=["Dimensi", "Butir"],
    how="left"
)

output_file = "hasil_uji_validitas_reliabilitas.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as w:
    gabung.to_excel(w, sheet_name="Validitas", index=False)
    tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)
    tabel_alpha_dihapus.to_excel(w, sheet_name="Alpha jika Dihapus", index=False)

print(f"File '{output_file}' berhasil dibuat.")
files.download(output_file)
```
